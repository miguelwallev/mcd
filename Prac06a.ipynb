````python
import numpy as np

def calcular_sincronia_ptp(t1, t2, t3, t4):
    """
    Simulación del cálculo de Offset y Delay en PTPv2.
    t1: Timestamp de envío Sync (Grandmaster)
    t2: Timestamp de recepción Sync (Esclavo)
    t3: Timestamp de envío Delay_Req (Esclavo)
    t4: Timestamp de recepción Delay_Req (Grandmaster)
    """
    # Ecuación de retardo del camino (asumiendo simetría)
    delay = ((t2 - t1) + (t4 - t3)) / 2

    # Ecuación de desfase del reloj esclavo
    offset = (t2 - t1) - delay

    return delay, offset

# Escenario: Grandmaster a 1ms, Red con 500ns de latencia y Esclavo adelantado 200ns
t1_base = 1000000 # nanosegundos
true_delay = 500
true_offset = 200

t2 = t1_base + true_delay + true_offset
t3 = 2000000 # Tiempo local esclavo
t4 = t3 + true_delay - true_offset

d, o = calcular_sincronia_ptp(t1_base, t2, t3, t4)

print(f"--- Resultados de Simulación PTPv2 ---")
print(f"Retardo del camino calculado: {d} ns")
print(f"Offset del reloj esclavo detectado: {o} ns")
print(f"Ajuste sugerido: {o * -1} ns")
````

### Descripción de la Función `calcular_sincronia_ptp`

La función `calcular_sincronia_ptp` simula el proceso central del protocolo PTPv2 (Precision Time Protocol version 2) para determinar la sincronización entre un reloj **Grandmaster** y un reloj **Esclavo**. Su propósito es calcular el retardo (delay) que experimentan los mensajes a través de la red y el desfase (offset) del reloj esclavo con respecto al Grandmaster. Estos dos valores son cruciales para que el reloj esclavo pueda ajustar su tiempo y sincronizarse con el Grandmaster.

#### Parámetros de Entrada:
La función toma cuatro marcas de tiempo (timestamps) como entrada, que representan momentos específicos en el intercambio de mensajes PTPv2 entre el Grandmaster y el Esclavo:

*   **`t1`**: Este parámetro representa el `Timestamp de envío Sync` del mensaje PTPv2. Es el momento en que el reloj **Grandmaster** envía un mensaje `Sync` al reloj Esclavo. Este timestamp se mide según el reloj del Grandmaster.

*   **`t2`**: Este parámetro representa el `Timestamp de recepción Sync`. Es el momento en que el reloj **Esclavo** recibe el mensaje `Sync` enviado por el Grandmaster. Este timestamp se mide según el reloj del Esclavo.

*   **`t3`**: Este parámetro representa el `Timestamp de envío Delay_Req`. Es el momento en que el reloj **Esclavo** envía un mensaje `Delay_Req` al Grandmaster en respuesta al mensaje `Sync`. Este timestamp se mide según el reloj del Esclavo.

*   **`t4`**: Este parámetro representa el `Timestamp de recepción Delay_Req`. Es el momento en que el reloj **Grandmaster** recibe el mensaje `Delay_Req` enviado por el Esclavo. Este timestamp se mide según el reloj del Grandmaster.

#### Ecuaciones de Cálculo:

La función utiliza las siguientes ecuaciones, basadas en los principios de PTPv2, para calcular el `delay` (retardo) y el `offset` (desfase):

1.  **Cálculo del Retardo del Camino (`delay`)**:
    ```python
    delay = ((t2 - t1) + (t4 - t3)) / 2
    ```
    *   **Explicación**: Esta ecuación calcula el retardo promedio de los mensajes entre el Grandmaster y el Esclavo. `(t2 - t1)` representa el tiempo que tardó el mensaje `Sync` en viajar del Grandmaster al Esclavo, medido por el reloj esclavo (aunque el inicio es por el grandmaster). `(t4 - t3)` representa el tiempo que tardó el mensaje `Delay_Req` en viajar del Esclavo al Grandmaster, medido por el reloj grandmaster. Sumando estos dos valores y dividiéndolos por dos, se obtiene un retardo de red promedio, asumiendo que el retardo en ambas direcciones es aproximadamente simétrico. Este retardo es crucial para aislar la diferencia horaria real entre los relojes de los efectos de la latencia de la red.

2.  **Cálculo del Desfase del Reloj Esclavo (`offset`)**:
    ```python
offset = (t2 - t1) - delay
    ```
    *   **Explicación**: Esta ecuación determina la diferencia de tiempo entre el reloj del Esclavo y el reloj del Grandmaster. `(t2 - t1)` es la diferencia de tiempo percibida por el Esclavo desde que el Grandmaster envió el `Sync` hasta que lo recibió. Esta diferencia incluye tanto el retardo de la red como cualquier desfase del reloj esclavo. Al restar el `delay` calculado previamente (que es la latencia de la red), el resultado es el `offset` puro del reloj del Esclavo con respecto al Grandmaster. Un `offset` positivo indica que el reloj del Esclavo está adelantado respecto al Grandmaster, mientras que un `offset` negativo indica que está atrasado. El esclavo utilizará este `offset` para ajustar su propio reloj.

*   **Incorporar la deriva del reloj del esclavo (drift) a lo largo del tiempo:**
    *   **Explicación:** Los osciladores de reloj no son perfectamente estables y su frecuencia puede variar ligeramente con el tiempo y las condiciones ambientales. Esta variación, conocida como deriva o _drift_, hace que el reloj del esclavo se acelere o ralentice respecto al _grandmaster_. Simular este `drift` (por ejemplo, añadiendo un pequeño cambio acumulativo al tiempo del esclavo entre mediciones) permitiría observar cómo el PTPv2 corrige continuamente este desfase dinámico y evaluar la eficacia de los mecanismos de ajuste a largo plazo.

*   **Añadir la consideración de errores de medición en los timestamps:**
    *   **Explicación:** Las mediciones de tiempo (`timestamps`) no son perfectas; pueden estar sujetas a ruido, latencia de interrupción o imprecisiones en el hardware.
